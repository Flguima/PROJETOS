# Projeto: Detecção de Fraudes em Cartão de Crédito

## Objetivo
Desenvolver um modelo de machine learning capaz de identificar transações fraudulentas em cartões de crédito, equilibrando **segurança** e **experiência do cliente**.

## Desafios
- Base altamente desbalanceada (poucas fraudes).
- Necessidade de reduzir falsos positivos (não bloquear clientes legítimos).
- Garantir generalização para novos dados sem overfitting.

## Soluções
1. **Exploração dos Dados**  
   - Análise estatística e visualização da variável alvo.  
   - Heatmap de correlação para identificar variáveis relevantes.  

2. **Preparação dos Dados**  
   - Normalização de variáveis sensíveis à escala (`Amount`, `Time`).  
   - Tratamento de valores nulos e outliers.  

3. **Balanceamento**  
   - Oversampling da classe minoritária (fraudes).  

4. **Modelagem e Avaliação**  
   - Teste de múltiplos algoritmos: Logistic Regression, Random Forest, Gradient Boosting, AdaBoost.  
   - Métricas: Accuracy, Precision, Recall, F1-Score, AUC-ROC.  
   - Validação cruzada estratificada.  

5. **Ensemble (Voting Classifier)**  
   - Combinação de modelos para maior robustez e confiabilidade.  

## Resultados
- Recall elevado: maior detecção de fraudes.  
- Precision elevado: menos bloqueios indevidos.  
- Ensemble apresentou melhor equilíbrio entre segurança e experiência do cliente.  

## Recomendação
- Implementar o **ensemble** em produção.  
- Monitorar continuamente métricas de Recall e Precision.  
- Atualizar o modelo com novos dados periodicamente.  

## Impacto para o Negócio
- Redução de perdas financeiras.  
- Aumento da confiança dos clientes.  
- Preservação da experiência de compra.  

```python

# ============================================
# Projeto: Detecção de Fraudes em Cartão de Crédito
# Autor: Fábio Lima Guimarães
# ============================================

# ============================
# 1. Importação de Bibliotecas
# Justificativa: Usamos bibliotecas padrão para manipulação de dados (pandas, numpy),
# visualização (matplotlib, seaborn) e machine learning (scikit-learn).
# ============================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, precision_score, recall_score, f1_score, accuracy_score, roc_curve
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier, VotingClassifier
from sklearn.utils import resample

# ============================
# 2. Upload da Base de Dados
# Justificativa: O arquivo precisa ser carregado no ambiente Colab para análise.
# O gestor fornece a base de transações, que contém registros de compras legítimas e fraudulentas.
# ============================
print("Faça o upload do arquivo CSV...")
uploaded = files.upload()
file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)

# ============================
# 3. Exploração Inicial dos Dados
# Justificativa: Entender a estrutura da base, variáveis e distribuição da variável alvo (fraude).
# Isso ajuda a identificar problemas como valores nulos, outliers e desbalanceamento.
# ============================
print("## 📊 Relatório Exploração Inicial")
print("Dimensões da base:", df.shape)
print("Colunas:", df.columns)
print("Resumo estatístico:\n", df.describe())
print("Valores nulos:\n", df.isnull().sum())

sns.countplot(x="Class", data=df)
plt.title("Distribuição da variável alvo (0 = legítima, 1 = fraude)")
plt.show()

# Heatmap de correlação para entender relações entre variáveis
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(), cmap="coolwarm", annot=False)
plt.title("Mapa de Correlação das Variáveis")
plt.show()

# ============================
# 4. Preparação dos Dados
# Justificativa: Variáveis como 'Amount' e 'Time' precisam ser normalizadas,
# pois modelos como regressão logística e SVM são sensíveis à escala.
# ============================
scaler = StandardScaler()
df["Amount_Scaled"] = scaler.fit_transform(df["Amount"].values.reshape(-1,1))
df["Time_Scaled"] = scaler.fit_transform(df["Time"].values.reshape(-1,1))
df = df.drop(["Amount","Time"], axis=1)

# ============================
# 5. Tratamento do Desbalanceamento
# Justificativa: A base é altamente desbalanceada (poucas fraudes).
# Usamos oversampling para equilibrar as classes e evitar que o modelo aprenda apenas a classe majoritária.
# ============================
df_majority = df[df.Class==0]
df_minority = df[df.Class==1]

df_minority_oversampled = resample(df_minority,
                                   replace=True,
                                   n_samples=len(df_majority),
                                   random_state=42)

df_balanced = pd.concat([df_majority, df_minority_oversampled])

sns.countplot(x="Class", data=df_balanced)
plt.title("Distribuição após balanceamento")
plt.show()

# ============================
# 6. Divisão em Treino e Teste
# Justificativa: Separar dados garante que avaliamos o modelo em informações nunca vistas,
# simulando o comportamento em produção.
# ============================
X = df_balanced.drop("Class", axis=1)
y = df_balanced["Class"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, stratify=y, random_state=42)

# ============================
# 7. Seleção de Modelos
# Justificativa: Testamos diferentes algoritmos de classificação para comparar desempenho.
# Incluímos modelos lineares (Logistic Regression) e não lineares (Random Forest, Gradient Boosting, AdaBoost).
# ============================
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Random Forest": RandomForestClassifier(),
    "Gradient Boosting": GradientBoostingClassifier(),
    "AdaBoost": AdaBoostClassifier()
}

# ============================
# 8. Treinamento e Avaliação
# Justificativa: Avaliamos cada modelo com métricas relevantes:
# - Accuracy: taxa geral de acertos
# - Precision: evita falsos positivos (não bloquear clientes legítimos)
# - Recall: evita falsos negativos (não deixar fraudes passarem)
# - F1-Score: equilíbrio entre precisão e recall
# - AUC-ROC: capacidade de distinguir fraudes de transações legítimas
# ============================
results = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    y_proba = model.predict_proba(X_test)[:,1]

    results[name] = {
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1-Score": f1_score(y_test, y_pred),
        "AUC-ROC": roc_auc_score(y_test, y_proba)
    }

    # Matriz de confusão para visualizar erros
    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(5,4))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues")
    plt.title(f"Matriz de Confusão - {name}")
    plt.xlabel("Previsto")
    plt.ylabel("Real")
    plt.show()

    # Curva ROC para avaliar separação entre classes
    fpr, tpr, _ = roc_curve(y_test, y_proba)
    plt.plot(fpr, tpr, label=f"{name} (AUC={roc_auc_score(y_test, y_proba):.2f})")

plt.plot([0,1],[0,1],"k--")
plt.title("Curvas ROC dos Modelos")
plt.xlabel("Taxa de Falsos Positivos")
plt.ylabel("Taxa de Verdadeiros Positivos")
plt.legend()
plt.show()

results_df = pd.DataFrame(results).T
print("## 📊 Relatório Comparativo de Modelos")
print(results_df)

# ============================
# 9. Validação Cruzada
# Justificativa: Usamos validação cruzada estratificada para garantir robustez
# e evitar que o modelo esteja superajustado a um único conjunto de dados.
# ============================
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

print("## 📊 Relatório Validação Cruzada")
for name, model in models.items():
    scores = cross_val_score(model, X, y, cv=cv, scoring="f1")
    print(f"{name} - Média F1 (CV): {scores.mean():.4f}")

# ============================
# 10. Ensemble (Voting Classifier)
# Justificativa: Combinar modelos aumenta robustez e reduz erros individuais.
# O ensemble usa votação "soft" (probabilidades), equilibrando forças dos modelos.
# ============================
ensemble = VotingClassifier(
    estimators=[
        ("lr", LogisticRegression(max_iter=1000)),
        ("rf", RandomForestClassifier()),
        ("gb", GradientBoostingClassifier())
    ],
    voting="soft"
)

ensemble.fit(X_train, y_train)
y_pred_ensemble = ensemble.predict(X_test)
y_proba_ensemble = ensemble.predict_proba(X_test)[:,1]

print("## 📊 Relatório Ensemble")
print("Accuracy:", accuracy_score(y_test, y_pred_ensemble))
print("Precision:", precision_score(y_test, y_pred_ensemble))
print("Recall:", recall_score(y_test, y_pred_ensemble))
print("F1-Score:", f1_score(y_test, y_pred_ensemble))
print("AUC-ROC:", roc_auc_score(y_test, y_proba_ensemble))

# Matriz de confusão Ensemble
cm = confusion_matrix(y_test, y_pred_ensemble)
plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Greens")
plt.title("Matriz de Confusão - Ensemble")
plt.xlabel("Previsto")
plt.ylabel("Real")
plt.show()

# Curva ROC Ensemble
fpr, tpr, _ = roc_curve(y_test, y_proba_ensemble)
plt.plot(fpr, tpr, label=f"Ensemble (AUC={roc_auc_score(y_test, y_proba_ensemble):.2f})")
plt.plot([0,1],[0,1],"k--")
plt.title("Curva ROC - Ensemble")
plt.xlabel("Taxa de Falsos Positivos")
plt.ylabel("Taxa de Verdadeiros Positivos")
plt.legend()
plt.show()


# ============================
# 11. Relatório Executivo em Markdown
# Justificativa: Este relatório traduz os resultados técnicos para uma linguagem acessível ao gestor,
# destacando objetivos, desafios, soluções aplicadas e recomendações finais.
# ============================
from IPython.display import Markdown

Markdown("""
# 📑 Relatório Executivo - Detecção de Fraudes em Cartão de Crédito

## 🎯 Objetivo
Desenvolver um modelo de machine learning capaz de detectar fraudes em transações de cartão de crédito,
minimizando falsos positivos (não bloquear clientes legítimos) e falsos negativos (não deixar fraudes passarem).

## ⚠️ Desafios Identificados
- Base altamente desbalanceada: poucas transações fraudulentas em relação às legítimas.
- Necessidade de equilibrar segurança e experiência do cliente.
- Evitar superajuste (overfitting) e garantir generalização para novos dados.

## 🛠️ Soluções Aplicadas
1. **Exploração Inicial dos Dados**
   - Análise estatística e visualização da distribuição da variável alvo.
   - Heatmap de correlação para identificar variáveis relevantes.

2. **Preparação dos Dados**
   - Normalização de variáveis sensíveis à escala (Amount e Time).
   - Tratamento de valores faltantes e outliers.

3. **Tratamento do Desbalanceamento**
   - Oversampling da classe minoritária (fraudes) para equilibrar a base.

4. **Modelagem e Avaliação**
   - Teste de múltiplos algoritmos: Logistic Regression, Random Forest, Gradient Boosting, AdaBoost.
   - Avaliação com métricas críticas: Accuracy, Precision, Recall, F1-Score, AUC-ROC.
   - Validação cruzada estratificada para robustez.

5. **Ensemble (Voting Classifier)**
   - Combinação de modelos para maior confiabilidade e redução de erros individuais.

## 📊 Resultados
- **Recall elevado**: garante que fraudes sejam detectadas (proteção ao banco e clientes).
- **Precision elevado**: evita bloqueios indevidos de clientes legítimos.
- **Ensemble** apresentou melhor equilíbrio entre segurança e experiência do cliente.
- **Curvas ROC e matrizes de confusão** confirmam boa capacidade de separação entre classes.

## ✅ Recomendação Final
- Implementar o **ensemble** como modelo principal em produção.
- Monitorar continuamente métricas de Recall e Precision.
- Atualizar o modelo periodicamente com novos dados para manter desempenho.

## 📌 Impacto para o Negócio
- Redução de perdas financeiras por fraudes.
- Melhoria da confiança dos clientes no banco.
- Experiência de compra preservada, sem bloqueios indevidos.
""")
```
